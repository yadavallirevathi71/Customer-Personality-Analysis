```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("marketing_campaign.csv",sep="\t")
```

```python
#Remove duplicates
df=df.drop_duplicates()
```

```python
#Handle invalid birth years
df.loc[df["Year_Birth"]<1900,"Year_Birth"]=np.nan
```

```python
# Handle extreme income
df.loc[df["Income"] > 200000, "Income"] = np.nan
```

```python
# Fill missing income with median
df["Income"] = df["Income"].fillna(df["Income"].median())
```

```python
# Standardize marital status
df["Marital_Status"] = df["Marital_Status"].replace({
    "Alone": "Single",
    "Absurd": "Single",
    "YOLO": "Single"
})
```

```python
# Convert date
df["Dt_Customer"] = pd.to_datetime(
    df["Dt_Customer"],
    format="%d-%m-%Y",
    errors="coerce"
)
```

```python
# Fill invalid birth years
df["Year_Birth"] = df["Year_Birth"].fillna(
    df["Year_Birth"].median()
).astype(int)
```

```python
# Remove constant columns
df = df.drop(columns=["Z_CostContact", "Z_Revenue"])

print("Dataset cleaned successfully!")
print("Shape:", df.shape)
print("Missing values:", df.isnull().sum().sum())
print("Duplicates:", df.duplicated().sum())
```
